### 실습 9.1: 배열과 기본 속성

1. 0부터 19까지의 정수를 갖는 1차원 배열 `array_1d`를 생성한다.
2. 자료형을 `int64`로 명시한다.
3. `shape`, `ndim`, `size`, `dtype`을 각각 변수에 저장한다.
4. `array_1d`를 `float64`로 변환한 새 배열 `array_float`를 생성한다.


````python
# TODO: None을 적절한 NumPy 표현식으로 바꾼다.
import numpy as np # 별칭

array_1d = np.arange(20, dtype=np.int64)
array_shape = array_1d.shape
array_ndim = array_1d.ndim
array_size = array_1d.size
array_dtype = array_1d.dtype
array_float = array_1d.astype(np.float64)
````

````python
task_1_passed = False
task_1_values = [
    array_1d,
    array_shape,
    array_ndim,
    array_size,
    array_dtype,
    array_float,
]

if any(value is None for value in task_1_values):
    print("TODO 9.1: 배열과 속성을 완성하세요.")
else:
    assert isinstance(array_1d, np.ndarray), "array_1d는 ndarray여야 합니다."
    assert np.array_equal(array_1d, np.arange(20, dtype=np.int64))
    assert array_shape == (20,)
    assert array_ndim == 1
    assert array_size == 20
    assert array_dtype == np.dtype("int64")
    assert np.array_equal(array_float, array_1d)
    assert array_float.dtype == np.dtype("float64")
    assert not np.shares_memory(array_1d, array_float)
    task_1_passed = True
    print("PASS 9.1")
````

### 실습 9.2: 배열 생성과 자료형 변환

1. 크기가 `(4, 5)`이고 모든 원소가 1인 배열 `ones_array`를 생성한다.
2. 크기가 `(3, 3)`이고 모든 원소가 0인 `int64` 배열 `zeros_array`를 생성한다.
3. `zeros_array`를 `float64`로 변환한 `zeros_float`를 생성한다.
4. 각 배열의 shape과 dtype을 직접 확인한다.


````python
# TODO: None을 적절한 NumPy 표현식으로 바꾼다.
ones_array = np.ones((4,5))
zeros_array = np.zeros((3,3),dtype=np.int64)
zeros_float = zeros_array.astype(np.float64)
````

````python
task_2_passed = False
task_2_values = [ones_array, zeros_array, zeros_float]

if any(value is None for value in task_2_values):
    print("TODO 9.2: 생성 함수와 자료형 변환을 완성하세요.")
else:
    assert isinstance(ones_array, np.ndarray)
    assert ones_array.shape == (4, 5)
    assert np.all(ones_array == 1)
    assert zeros_array.shape == (3, 3)
    assert zeros_array.dtype == np.dtype("int64")
    assert np.all(zeros_array == 0)
    assert zeros_float.shape == (3, 3)
    assert zeros_float.dtype == np.dtype("float64")
    assert not np.shares_memory(zeros_array, zeros_float)
    task_2_passed = True
    print("PASS 9.2")
````

### 실습 9.3: 벡터화와 브로드캐스팅

다음 배열을 사용한다.

```python
base = np.array([[1, 2, 3], [4, 5, 6]])
```

1. 모든 원소를 제곱한 `squared`를 계산한다.
2. 모든 원소에 10을 곱한 `scaled`를 계산한다.
3. 열별 보정값 `[10, 20, 30]`을 `adjustment`로 만든다.
4. `base`와 `adjustment`를 더한 `adjusted`를 계산한다.
5. Python 반복문을 사용하지 않는다.


````python
base = np.array([[1, 2, 3], [4, 5, 6]])

# TODO: None을 벡터화된 NumPy 표현식으로 바꾼다.
squared = base**2
scaled = base*10
adjustment = np.array([10,20,30])
adjusted = base+adjustment
````

````python
task_3_passed = False
task_3_values = [squared, scaled, adjustment, adjusted]

if any(value is None for value in task_3_values):
    print("TODO 9.3: 벡터화와 브로드캐스팅을 완성하세요.")
else:
    assert np.array_equal(squared, np.array([[1, 4, 9], [16, 25, 36]]))
    assert np.array_equal(scaled, np.array([[10, 20, 30], [40, 50, 60]]))
    assert np.array_equal(adjustment, np.array([10, 20, 30]))
    assert adjustment.shape == (3,)
    assert np.array_equal(adjusted, np.array([[11, 22, 33], [14, 25, 36]]))
    task_3_passed = True
    print("PASS 9.3")
````

### 실습 9.4: View와 Copy

1. `np.arange(20)`으로 `source`를 생성한다.
2. `source[5:10]`을 `view`에 저장하고 모든 값을 -1로 바꾼다.
3. 별도의 `copy_source`를 생성한다.
4. `copy_source[5:10].copy()`를 `independent`에 저장하고 모든 값을 -1로 바꾼다.
5. view를 바꿀 때와 copy를 바꿀 때 원본이 어떻게 달라지는지 비교한다.


````python
# TODO: 순서대로 배열을 만들고 선택한 구간을 변경한다.
source = np.arange(20)
view = source[5:10]
view[:]=-1

copy_source = np.arange(20)
independent = copy_source[5:10].copy()
independent[:]=-1
````

````python
task_4_passed = False
task_4_values = [source, view, copy_source, independent]

if any(value is None for value in task_4_values):
    print("TODO 9.4: view와 copy 비교를 완성하세요.")
else:
    expected_changed = np.arange(20)
    expected_changed[5:10] = -1
    assert np.array_equal(source, expected_changed)
    assert np.array_equal(view, np.full(5, -1))
    assert np.shares_memory(source, view)
    assert np.array_equal(copy_source, np.arange(20))
    assert np.array_equal(independent, np.full(5, -1))
    assert not np.shares_memory(copy_source, independent)
    task_4_passed = True
    print("PASS 9.4")
````

### 실습 9.5: 2차원 인덱싱

`np.arange(1, 13).reshape(3, 4)`로 `matrix`를 생성한다.

1. 두 번째 행 전체를 `second_row`에 저장한다.
2. 세 번째 열 전체를 `third_column`에 저장한다.
3. 좌상단 2×2 영역을 `top_left`에 저장한다.
4. 두 번째 행을 2차원 형태로 유지하여 `second_row_2d`에 저장한다.


````python
# TODO: None을 적절한 인덱싱 또는 슬라이싱 표현식으로 바꾼다.
matrix = np.arange(1,13).reshape(3,4)
second_row = matrix[1]
third_column = matrix[:,2]
top_left = matrix[:2,:2]
second_row_2d = matrix[1:2]
````

````python
task_5_passed = False
task_5_values = [matrix, second_row, third_column, top_left, second_row_2d]

if any(value is None for value in task_5_values):
    print("TODO 9.5: 2차원 인덱싱을 완성하세요.")
else:
    expected_matrix = np.arange(1, 13).reshape(3, 4)
    assert np.array_equal(matrix, expected_matrix)
    assert np.array_equal(second_row, np.array([5, 6, 7, 8]))
    assert second_row.shape == (4,)
    assert np.array_equal(third_column, np.array([3, 7, 11]))
    assert third_column.shape == (3,)
    assert np.array_equal(top_left, np.array([[1, 2], [5, 6]]))
    assert top_left.shape == (2, 2)
    assert np.array_equal(second_row_2d, np.array([[5, 6, 7, 8]]))
    assert second_row_2d.shape == (1, 4)
    task_5_passed = True
    print("PASS 9.5")
````

### 실습 9.6: Boolean 조건 처리

주어진 `data` 배열을 사용한다.

1. 음수 위치를 나타내는 `negative_mask`를 만든다.
2. 음수 원소만 `negative_values`에 저장한다.
3. 원본을 복사한 `cleaned`에서 음수를 0으로 바꾼다.
4. 0 이상 3 이하인 위치를 나타내는 `middle_mask`를 만든다.
5. `np.where`로 음수를 0으로 바꾼 새 배열 `replaced`를 만든다.


````python
data = np.array([
    [4, -2, 1],
    [-1, 0, 3],
    [2, -5, 6],
])

# TODO: None을 Boolean 연산과 NumPy 선택 표현식으로 바꾼다.
negative_mask = data<0
negative_values = data[negative_mask]

cleaned = data.copy()
cleaned[negative_mask]=0
middle_mask = (data>=0) & (data<=3)
replaced = np.where(data<0,0,data)
````

````python
task_6_passed = False
task_6_values = [negative_mask, negative_values, cleaned, middle_mask, replaced]

if any(value is None for value in task_6_values):
    print("TODO 9.6: Boolean 조건 처리를 완성하세요.")
else:
    expected_cleaned = np.array([[4, 0, 1], [0, 0, 3], [2, 0, 6]])
    expected_middle = np.array(
        [[False, False, True], [False, True, True], [True, False, False]]
    )
    assert negative_mask.dtype == np.dtype("bool")
    assert np.array_equal(negative_mask, data < 0)
    assert np.array_equal(negative_values, np.array([-2, -1, -5]))
    assert np.array_equal(cleaned, expected_cleaned)
    assert np.array_equal(data, np.array([[4, -2, 1], [-1, 0, 3], [2, -5, 6]]))
    assert np.array_equal(middle_mask, expected_middle)
    assert np.array_equal(replaced, expected_cleaned)
    task_6_passed = True
    print("PASS 9.6")
````

### 실습 9.7: 형태 변경과 Advanced Indexing

1. 1부터 12까지의 값을 갖는 `sequence`를 생성한다.
2. 이를 `(3, 4)` 형태의 `matrix_3x4`로 변경한다.
3. 전치한 결과를 `transposed`에 저장한다.
4. 행 순서 `[2, 0]`을 선택한 결과를 `reordered`에 저장한다.
5. `reordered`의 첫 원소를 -99로 바꾸고 원본이 유지되는지 확인한다.


````python
# TODO: None을 형태 변경과 정수 배열 인덱싱 표현식으로 바꾼다.
sequence = np.arange(1,13)
matrix_3x4 = sequence.reshape(3,4)
transposed = matrix_3x4.T
reordered = matrix_3x4[[2,0]]
reordered[0,0]=-99
````

````python
task_7_passed = False
task_7_values = [sequence, matrix_3x4, transposed, reordered]

if any(value is None for value in task_7_values):
    print("TODO 9.7: 형태 변경과 advanced indexing을 완성하세요.")
else:
    expected_matrix = np.arange(1, 13).reshape(3, 4)
    assert np.array_equal(sequence, np.arange(1, 13))
    assert np.array_equal(matrix_3x4, expected_matrix)
    assert matrix_3x4.shape == (3, 4)
    assert np.array_equal(transposed, expected_matrix.T)
    assert transposed.shape == (4, 3)
    assert reordered.shape == (2, 4)
    assert reordered[0, 0] == -99
    assert np.array_equal(reordered[1], expected_matrix[0])
    assert matrix_3x4[2, 0] == 9
    assert not np.shares_memory(matrix_3x4, reordered)
    task_7_passed = True
    print("PASS 9.7")
````

### 실습 9.8: 난수와 축별 집계

1. seed가 42인 기본 난수 생성기 `rng`을 만든다.
2. 50 이상 101 미만의 정수로 `(5, 4)` 배열 `scores`를 생성한다.
3. 전체 평균을 `overall_mean`에 저장한다.
4. 열별 평균을 `column_means`에 저장한다.
5. 행별 표준편차를 `row_std`에 저장한다.
6. 각 행의 모든 점수가 60 이상인지 나타내는 `all_passed`를 만든다.


````python
# TODO: None을 난수 생성과 축별 집계 표현식으로 바꾼다.
rng = np.random.default_rng(seed=42)
scores = rng.integers(50,101,size=(5,4))
overall_mean = scores.mean()
column_means = scores.mean(axis=0)
row_std = scores.std(axis=1)
all_passed = (scores>=60).all(axis=1)
````

````python
task_8_passed = False
task_8_values = [rng, scores, overall_mean, column_means, row_std, all_passed]

if any(value is None for value in task_8_values):
    print("TODO 9.8: 난수와 축별 집계를 완성하세요.")
else:
    expected_rng = np.random.default_rng(seed=42)
    expected_scores = expected_rng.integers(50, 101, size=(5, 4))
    assert isinstance(rng, np.random.Generator)
    assert np.array_equal(scores, expected_scores)
    assert scores.shape == (5, 4)
    assert scores.min() >= 50 and scores.max() < 101
    assert np.isscalar(overall_mean)
    assert np.isclose(overall_mean, expected_scores.mean())
    assert np.allclose(column_means, expected_scores.mean(axis=0))
    assert column_means.shape == (4,)
    assert np.allclose(row_std, expected_scores.std(axis=1))
    assert row_std.shape == (5,)
    assert np.array_equal(all_passed, (expected_scores >= 60).all(axis=1))
    assert all_passed.shape == (5,)
    task_8_passed = True
    print("PASS 9.8")
````
